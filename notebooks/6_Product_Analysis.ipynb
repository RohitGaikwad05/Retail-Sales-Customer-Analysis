```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("transformed_retail_sales.csv")
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
print(df)
```

Top Products by Sales

```python
product_sales = df.groupby("Product Name")["Sales"].sum()

top_10_products = product_sales.sort_values(ascending=False).head(10)

print(top_10_products)
```

Bottom Products

```python
bottom_10_products = product_sales.sort_values(ascending=True).head(10)

print(bottom_10_products)
```

Most Profitable Products

```python
product_profit = df.groupby("Product Name", as_index=False)["Profit"].sum()

most_profitable_products = product_profit.sort_values("Profit", ascending=False).head(10)

print(most_profitable_products)
```

Low-Margin Products

```python
product_performance = df.groupby("Product Name", as_index=False).agg({
    "Sales": "sum",
    "Profit": "sum"
})

product_performance["Profit Margin"] = (
    product_performance["Profit"] /
    product_performance["Sales"]
) * 100

print(product_performance.head())
```

```python
high_sales_limit = product_performance["Sales"].median()

low_margin_products = product_performance[
    (product_performance["Sales"] > high_sales_limit) &
    (product_performance["Profit Margin"] < 10)
]

low_margin_products = low_margin_products.sort_values(
    "Sales", ascending=False
)

print(low_margin_products)
```

Declining Products

```python
df["Year_Month"] = df["Order Date"].dt.to_period("M").astype(str)

monthly_product_sales = df.groupby(
    ["Product Name", "Year_Month"], as_index=False
)["Sales"].sum()

print(monthly_product_sales.head(10))
```

```python
monthly_product_sales["Previous Month Sales"] = (
    monthly_product_sales.groupby("Product Name")["Sales"].shift(1)
)

monthly_product_sales["Sales Change"] = (
    monthly_product_sales["Sales"] -
    monthly_product_sales["Previous Month Sales"]
)

print(monthly_product_sales.head(10))
```

```python
monthly_product_sales["Decreased"] = (
    monthly_product_sales["Sales Change"] < 0
)

monthly_product_sales["Decrease Group"] = (
    monthly_product_sales.groupby("Product Name")["Decreased"]
    .transform(
        lambda x: (x != x.shift()).cumsum()
    )
)

declining_products = monthly_product_sales[
    monthly_product_sales["Decreased"]
].copy()

declining_products["Consecutive Decreases"] = (
    declining_products.groupby(
        ["Product Name", "Decrease Group"]
    )["Decreased"].transform("sum")
)

declining_products = declining_products[
    declining_products["Consecutive Decreases"] >= 3
]

print(
    declining_products[
        ["Product Name", "Year_Month", "Sales", "Sales Change"]
    ]
)
```

Category Performance

```python
category_performance = df.groupby("Category", as_index=False).agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "nunique"
})

category_performance["Profit Margin"] = (category_performance["Profit"] /category_performance["Sales"]) * 100

category_performance = category_performance.rename(columns={"Order ID": "Orders"})

print(category_performance)
```
