```python
import pandas as pd
```

```python
df= pd.read_csv("Cleaned_superstore_dataset2011-2015.csv")
```

```python
print(df["Order Date"].dtype)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Year"] = df["Order Date"].dt.year

df["Month"] = df["Order Date"].dt.month

df["Month_Name"] = df["Order Date"].dt.month_name()

df["Quarter"] = df["Order Date"].dt.quarter
```

```python
print(df[["Order Date","Year","Month","Month_Name","Quarter"]].head())
```

The .dt accessor in pandas is a special property used with datetime columns (Series). It allows you to access various datetime-specific properties and methods.

Profit Margin

```python
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100
```

```python
print(df[["Sales","Profit","Profit_Margin"]].head())
```

Order Value

```python
df["Order_Value"] = df["Sales"]
print(df["Order_Value"].head())
```

Customer Segmentation

```python
customer_sales = df.groupby("Customer ID")["Sales"].sum()

print(customer_sales.head())
```

```python
def customer_segment(sales):

    if sales >= 5000:
        return "High Value"

    elif sales >= 2000:
        return "Medium Value"

    else:
        return "Low Value"
```

```python
customer_segment = customer_sales.apply(customer_segment)
```

```python
df["Customer_Segment"] = df["Customer ID"].map(customer_segment)
```

```python
print(df[["Customer ID","Sales","Customer_Segment"]].head())
```

Sales Category

```python
def sales_category(sales):

    if sales >= 1000:
        return "High Sales"

    elif sales >= 500:
        return "Medium Sales"

    else:
        return "Low Sales"
```

```python
df["Sales_Category"] = df["Sales"].apply(sales_category)
```

```python
print(df[["Sales","Sales_Category"]].head())
```

Profit Category


```python
def profit_category(profit):

    if profit > 0:
        return "Profitable"

    elif profit < 0:
        return "Loss"

    else:
        return "No Profit"


df["Profit_Category"] = (
    df["Profit"].apply(profit_category)
)
```

```python
print(df[['Profit','Profit_Category']].head())
```

Customer Type

```python
customer_orders = df.groupby("Customer ID")["Order ID"].nunique()

def customer_type(orders):

    if orders > 1:
        return "Repeat Customer"

    else:
        return "One-Time Customer"


customer_type_result = (
    customer_orders.apply(customer_type)
)
```

```python
df["Customer_Type"] = (df["Customer ID"].map(customer_type_result))
```

```python
print(df[['Customer ID','Customer_Type']].head())
```

Product Performance

```python
def product_performance(profit):

    if profit > 0:
        return "Good"

    else:
        return "Poor"


df["Product_Performance"] = (df["Profit"].apply(product_performance))
```

```python
print(df[['Profit','Product_Performance']].head())
```

Order Size

```python
def order_size(quantity):

    if quantity >= 5:
        return "Large"

    elif quantity >= 2:
        return "Medium"

    else:
        return "Small"


df["Order_Size"] = (df["Quantity"].apply(order_size))
```

```python
print(df[['Quantity','Order_Size']].head())
```

```python
df.to_csv("transformed_retail_sales.csv", index=False)
```
