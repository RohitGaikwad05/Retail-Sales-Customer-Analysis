```python
import pandas as pd
```

```python
df= pd.read_csv("superstore_dataset2011-2015.csv",encoding="ISO-8859-1")
print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
df = df.drop("Postal Code", axis=1)
```

Postal Code contained a large number of missing values.

Mean or median filling was not appropriate because Postal Code
represents a location identifier.

Since Postal Code was not
required for the main business analysis, the column was removed.

```python
print("before rows:",len(df))
duplicates= df.duplicated().sum()
print("duplicates records:",duplicates)

df=df.drop_duplicates()


print("after rows:",len(df))
```

No duplicate records were found, therefore no records were
removed during duplicate cleaning.

```python
print(df.dtypes)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed", dayfirst=True)
print(df['Order Date'].dtype)
```

```python
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="mixed", dayfirst=True)
print(df["Ship Date"].dtype)
```

```python
df["Sales"] = pd.to_numeric(df["Sales"])
print(df["Sales"].dtype)
```

```python
print(df.dtypes)
```

```python
negative_sales = df[df["Sales"] < 0]
print(negative_sales)
```

No negative sales values were found, so no sales records
were removed.
